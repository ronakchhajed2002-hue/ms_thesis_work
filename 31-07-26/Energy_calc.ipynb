```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Functions
def Energy_theta_n(theta_n):
    E_theta_n = np.sum(np.cos(theta_n))
    return E_theta_n
```

```python
# Ploting Energy vs theta
# considering  J>0 that

#number of sides of a polygon, n_gon > 0
n_gon = 6

#Case with angles 180,180,180......,-(n_gon-1)*180
theta_n = np.zeros(n_gon) + np.array([180])
theta_n[n_gon - 1] = 0
theta_n[n_gon - 1] = (-1) * np.sum(theta_n)
theta_n = np.radians(theta_n)
energy = Energy_theta_n(theta_n)
print("For case with angles 180,180,180......,-(n_gon-1)*180")
print(f"---> Energy / J = {energy}")

theta_d = np.linspace(0, 360, 500)
theta_r = np.radians(theta_d)

#Symmetry of the problem allows all thetas to be equal to find the minima
E_theta = (n_gon -1) * np.cos(theta_r) + np.cos((n_gon -1) * theta_r)

#plotting energy vs theta
xticks = np.linspace(0, 360, n_gon*4 + 1)
yticks = (n_gon -1) * np.cos(xticks) + np.cos((n_gon -1) * xticks)
plt.plot(theta_d, E_theta)
plt.xlabel("Angle theta in degrees")
plt.ylabel("Energy / J")
plt.xticks(xticks)
plt.scatter(180, energy, label="energy corresponding to theta = 180")
plt.axhline(y=energy, linestyle=':')
plt.axvline(x=180, linestyle=':')
plt.legend(loc='upper right')
plt.savefig('energy_plot.png', dpi=300)
plt.show()
```

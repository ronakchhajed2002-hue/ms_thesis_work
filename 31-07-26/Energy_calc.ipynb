```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Functions

def Energy_theta(theta_1, theta_2, theta_3):
    E_theta = np.cos(theta_1) + np.cos(theta_2) + np.cos(theta_3)
    return E_theta
```

```python
# Ploting Energy vs theta
# considering  J>0 that

theta_d = np.linspace(0, 360, 50)
theta_r = np.radians(theta_d)
E_theta = 2 * np.cos(theta_r) + np.cos(2 * theta_r)
plt.scatter(theta_d, E_theta)
plt.xlabel("Angle theta in degrees")
plt.ylabel("Energy / J")
plt.savefig('energy_plot.png', dpi=300)
plt.show()

#Case with angles 0, 180, 180

theta_1 = np.radians(0)
theta_2 = np.radians(180)
theta_3 = np.radians(180)

energy = Energy_theta(theta_1, theta_2, theta_3)
print(f"Energy / J = {energy}")
```

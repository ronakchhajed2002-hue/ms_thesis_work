```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Functions

def E_o(theta_1, theta_2, theta_3):
    E = np.cos(theta_1) + np.cos(theta_2) + np.cos(theta_3)
    return E
```

```python
# Ploting Energy vs theta
theta_d = np.linspace(0, 360, 50)
theta_r = np.radians(theta_d)
E = 2 * np.cos(theta_r) + np.cos(2 * theta_r)
plt.scatter(theta_d, E)
plt.xlabel("Angle theta")
plt.ylabel("Energy / J")
plt.savefig('my_plot1.png', dpi=300)
plt.show()

#Case with angles 0, 180, 180
theta_1 = np.radians(0)
theta_2 = np.radians(180)
theta_3 = np.radians(180)

Energy = E_o(theta_1, theta_2, theta_3)
print(f"Energy = {Energy}")
```
